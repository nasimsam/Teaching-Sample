<font color="#FF7500" size="6"><div id="data_structures">Data Structures</div><hr></font>
<p></p>
Data structures, as their name suggests, are methods—or, in other words, structures—used to store data in an organized way so that our queries can be answered efficiently and quickly. If the data represented by a data structure changes frequently, the data structure should also be able to adapt quickly to those changes.
<br>
Clearly, depending on what questions we ask about a set of data, how much the data may change (for example, whether elements are deleted or added), or what kind of changes occur (for example, whether only the smallest element is removed or any element can be removed), the way the data should be stored must also be different.
<br>
The definitions above may seem somewhat vague. Consider the following example. Suppose Aria has a box containing screws of different sizes. Every time Milad gives her a nut, she wants to know whether she has a screw with the same size. Aria uses an array to store the sizes of the screws. She has two approaches:
<ul>
<li>
Store the sizes in the array without any particular order. The advantage is that when Aria buys a new screw, she can easily append its size to the end of the array. However, whenever she receives a nut, she must use a loop to check the entire array to determine whether a screw of the same size exists.

</li>
<li>
Store the screw sizes in sorted order. When she receives a new nut, she can use binary search to find the matching screw much faster and more efficiently than checking every array element. The drawback is that if Aria buys a new screw, inserting its size while preserving the sorted order can be difficult and time-consuming.

<font color="#FF7500" size="6"><div id="queue">Queue</div><p></p></font>
<hr>
A queue is a data structure in which a new element is added only at the end, and an element is removed only from the front. It is like a bakery line: each person who arrives goes to the end of the line, and each person leaves from the front after receiving their bread.
<br><p></p>
A queue should support the following operations:
<ul>
<li>$enqueue(x)$: inserts element $x$ at the end of the queue.</li>
<li>$dequeue()$: removes and returns the first element.</li>
<li>$front()$: returns the first element of the queue.</li>
<li>$size()$: returns the number of elements in the queue.</li>
<li>$is\_empty()$: determines whether the queue is empty.</li>
<li>$is\_full()$: determines whether the queue is full.</li>
</ul>
<b>Note</b>: All of these operations are performed in $O(1)$ time.
<p></p>
A queue is usually implemented using an array or a linked list, which we will learn about later.
<p></p>


<font size="6">Implementing a Queue Using an Array</font>
<p></p>
To implement a queue using an array, we first define an array and two variables: one representing the beginning of the queue ($first$), and another representing the number of elements currently in the queue ($num$).
<br>
When a new element is added, we place it at the end of the queue and then increase $num$ by one. To remove the first element of the queue, we decrease $num$ by one and increase $first$ by one.
<br>
(Note: In this explanation, we assume that the array is used circularly. For example, if the array size is 5, the element after the fifth position is the first position.)


```python
class Queue:
    def __init__(self, max_size):
        self.max_size = max_size
        self.Q = [0] * max_size
        self.num = 0
        self.first = 0
    
    def enqueue(self, item):
        if self.num >= self.max_size:
            raise Exception("Queue overflow")
        self.Q[(self.num + self.first) % self.max_size] = item
        self.num += 1
        
    def dequeue(self):
        if self.num == 0:
            raise Exception("Queue empty")
        item = self.Q[self.first]
        self.first = (self.first + 1) % self.max_size
        self.num -= 1
        return item
    
    def front(self):
        if self.num == 0:
            raise Exception("Queue empty")
        return self.Q[self.first]
    
    def is_empty(self):
        return self.num == 0
    
    def size(self):
        return self.num
    
    def is_full(self):
        return self.num >= self.max_size
```

```python
# Example
q = Queue(10) # (front of queue)[](back of queue)
q.enqueue("ra'na") # ["ra'na"]
q.enqueue("vez") # ["ra'na", "vez"]
q.enqueue("Arya") # ["ra'na", "vez", "Arya"]
print("Queue size is:", q.size())
print(q.dequeue(), "left the queue.") # ["vez", "Arya"]
print("Front of queue is:", q.front())
q.enqueue("milda") # ["vez", "Arya", "milda"]
q.dequeue() # ["Arya","milda"]
q.dequeue() # ["milda"]
q.dequeue() # []
print("It was a queue!")
```

<font size="6" color="#FF7500"><div id="stack">Stack</div><p></p></font>
<hr>
In some situations, elements are added from only one side and are also removed from that same side. For example, consider a stack of dirty dishes that you want to wash. Whenever you want to remove a dish, you select the top dish and wash it. If you want to add a dirty dish, you place it on top of all the others.
<br>
In other words, the element that was added most recently is removed first. A data structure that models this behavior is called a stack.
<p></p>
A stack should support the following functions:
<ul>
<li>$push(x)$: adds $x$ to the top of the stack.</li>
<li>$pop()$: removes and returns the top element of the stack.</li>
<li>$top()$: returns the top element of the stack.</li>
<li>$size()$: returns the number of elements in the stack.</li>
<li>$is\_empty()$: determines whether the stack is empty.</li>
<li>$is\_full()$: determines whether the stack is full.</li>
</ul>
<b>Note</b>: All of the above operations are performed in $O(1)$ time.
<p></p>
A stack can also be implemented using an array or a linked list.


<font size="6">Implementing a Stack Using an Array</font>
<p></p>
To implement a stack using an array, we use a variable $num$ that always indicates the top of the stack.


```python
class Stack:
    def __init__(self, max_size):
        self.max_size = max_size  # Size of stack
        self.S = [0] * max_size  # Stack array
        self.num = 0  # Number of elements in Stack
    
    def push(self, item):
        if self.num >= self.max_size:
            raise Exception("Stack overflow")
        self.S[self.num] = item
        self.num += 1
        
    def pop(self):
        if self.num == 0:
            raise Exception("Stack empty")
        self.num -= 1
        return self.S[self.num]
    
    def top(self):
        if self.num == 0:
            raise Exception("Stack empty")
        return self.S[self.num-1]

    
    def size(self):
        return self.num
    
    def is_full(self):
        return self.num >= self.max_size
    
    def is_empty(self):
        return self.num == 0
```

```python
# Example
st = Stack(10) # (top of stack)[]
st.push(10) # [10]
st.push(30) # [30, 10]
st.push(1.5) # [1.5, 30, 10]
print(st.pop(), "is popped.") # [30, 10]
st.push("salam") # ["salam", 30, 10]
st.pop() # [30, 10]
print("Top of stack:", st.top())
st.pop() # [10]
st.pop() # []
print("Stack is empty?", st.is_empty())
```

<font size="6" color="#FF7500"><div id="deque">Deque (Double-Ended Queue)</div><p></p></font>
<hr>
A regular queue has limitations when modeling certain situations. For example, suppose we are modeling a bakery line, but the baker puts his acquaintances at the front of the line!
Therefore, sometimes we need to add an element to the front of the queue, or even remove an element from the back of the queue. For example, someone who does not want to wait might convince several people at the end of the line to leave and then sneak into the end of the line!
In fact, a deque supports the capabilities of both a stack and a queue at the same time.
<p></p>
A deque should support the following functions:
<ul>
<li>$push\_front(x)$: inserts element $x$ at the front of the deque.</li>
<li>$push\_back(x)$: inserts element $x$ at the back of the deque.</li>
<li>$pop\_front(x)$: removes and returns the element at the front of the deque.</li>
<li>$pop\_back(x)$: removes and returns the element at the back of the deque.</li>
<li>$front()$: returns the element at the front of the deque.</li>
<li>$back()$: returns the element at the back of the deque.</li>
<li>$size()$: returns the number of elements in the deque.</li>
<li>$is\_empty()$: determines whether the deque is empty.</li>
<li>$is\_full()$: determines whether the deque is full.</li>
</ul>
<b>Note</b>: All of the above operations are performed in $O(1)$ time.
<p></p>
A deque can also be implemented using an array or a linked list.


<font size="6">Implementing a Deque Using an Array</font>
<p></p>
Implementing a deque with an array is similar to implementing a regular queue using an array.
The only difference is that when an element is added to the front of the queue, we decrease the value of $first$ by one and increase $num$ by one.
When an element is removed from the back of the queue, we only decrease $num$ by one.


```python
class DoubleEndedQueue:
    def __init__(self, max_size):
        self.max_size = max_size
        self.Q = [0] * max_size
        self.num = 0
        self.first = 0
    
    def push_back(self, item): # Like one-way queue
        if self.num >= self.max_size:
            raise Exception("Queue overflow")
        self.Q[(self.num + self.first) % self.max_size] = item
        self.num += 1

    def push_front(self, item): # New
        if self.num >= self.max_size:
            raise Exception("Queue overflow")
        self.first = (self.first - 1) % self.max_size
        self.Q[self.first] = item
        self.num += 1

    def pop_front(self): # Like one-way queue
        if self.num == 0:
            raise Exception("Queue empty")
        item = self.Q[self.first]
        self.first = (self.first + 1) % self.max_size
        self.num -= 1
        return item
    
    def front(self): # Like one-way queue
        if self.num == 0:
            raise Exception("Queue empty")
        return self.Q[self.first]
        
    def pop_back(self): # New
        if self.num == 0:
            raise Exception("Queue empty")
        self.num -= 1
        return self.Q[(self.num + self.first) % self.max_size]
    
    def back(self): # New
        if self.num == 0:
            raise Exception("Queue empty")
        return self.Q[(self.num + self.first - 1) % self.max_size]
    
    def is_empty(self):
        return self.num == 0
    
    def size(self):
        return self.num
    
    def is_full(self):
        return self.num >= self.max_size
```

```python
# Example
dq = DoubleEndedQueue(7) # (front of queue)[](back of queue)
dq.push_front(1) # [1]
dq.push_back(2) # [1, 2]
dq.push_front(4) # [4, 1, 2]
print("Queue size is:", dq.size())
dq.push_back(0) # [4, 1, 2, 0]
print("Back of queue is:", dq.back())
dq.pop_front() # [1, 2, 0]
print("Front of queue is:", dq.front())
dq.pop_back() # [1, 2]
dq.pop_back() # [1]
dq.pop_back() # []
print("Queue size is:", dq.size())
```

<font size="6" color="#FF7500"><div id="linked_list">Linked List</div><p></p></font>
<hr>
A linked list is a very basic data structure for storing sequences. In a linked list, each element knows only about its next element. In a doubly linked list, each element also knows about its previous element.
<br>
For the first and last elements, there are two possible approaches: one is to have no element before the first or after the last; the other is to make the linked list circular, so that the element after the last is the first element and the element before the first is the last element.
<p></p>
A linked list supports the following functions:
<ul>
<li>$insert\_after(data,x)$: inserts element $data$ after element $x$.</li>
<li>$delete(x)$: removes and returns element $x$.</li>
<li>$find(value)$: returns the first element whose $data$ equals $value$.</li>
<li>$get(i)$: returns the $i$-th element of the list.</li>
<li>$size()$: returns the number of elements in the list.</li>
<li>$is\_empty()$: determines whether the list is empty.</li>
</ul>
<p></p>
Consider the following table:
<table>
<thead>
<tr><th style="text-align:center;">Operation</th><th style="text-align:center;">Array</th><th style="text-align:center;">Linked List</th></tr>
</thead>
<tbody>
<tr><td style="text-align:center;">Insertion at a specified position and deletion</td><td style="text-align:center;">$O(n)$</td><td style="text-align:center;">search time + $O(1)$</td></tr>
<tr><td style="text-align:center;">Accessing the $i$-th element</td><td style="text-align:center;">$O(1)$</td><td style="text-align:center;">$O(n)$</td></tr>
</tbody>
</table>
Insertion at a position in an array is $O(n)$ because the elements after the insertion position must be shifted, and there can be $O(n)$ such elements. However, to insert an element such as $x$ at a specific position in a linked list, such as position $i$, it is enough to set the next pointer of the $(i-1)$-th element to $x$ and the next pointer of $x$ to the old $i$-th element.
Deletion works similarly.
We have direct access to the $i$-th element in an array, but in a linked list we must move forward from the first element $i$ times, so this operation is $O(n)$.


<font size="6"><br>Implementing a Linked List Using Classes</font>
<br>
We define a class for the nodes of our list. For each $node$, we store the required information in $data$. We also store the node after and before it in $next$ and $prev$, respectively.
Then, using a $List$ class in which $head$ serves as the node before the first element and after the last element, we implement a doubly linked list.
<br>
In fact, $head$ is a $node$ whose value is $None$, and its purpose is to keep track of both the beginning and the end of our linked list.


```python
class Node:
    def __init__(self, data):
        self.data = data  
        self.next = None  
        self.prev = None  
    
class List:
    def __init__(self):        
        self.head = Node(None)
        self.head.next = self.head
        self.head.prev = self.head
        self.n = 0
        
    def get(self,ind):
        if ind >= self.size() : 
            raise Exception("Out of list")
        x = self.head.next
        for i in range(ind) :
            x = x.next
        return x
    
    def insert_after(self, x, data):
        y = Node(data)
        self.n += 1
        y.prev = x
        y.next = x.next
        x.next = y
        y.next.prev = y
        return y
        
    def delete(self, x):
        self.n -= 1
        x.prev.next = x.next
        x.next.prev = x.prev
        return x
        
    def find(self, val):
        x = self.head.next
        for i in range(self.size()) :
            if x.data == val :
                return x
            x = x.next
        return None
    
    def size(self):
        return self.n
      
    def is_empty(self):
        return self.n==0
```

```python
# Example
list = List() # head
list.insert_after(list.head, "milad") # head <-> milad
list.insert_after(list.get(0), "Arya") # head <-> milad <-> Arya
list.insert_after(list.find("Arya").prev, "jabbar") # head <-> milad <-> jabbar <-> Arya 
list.delete(list.find("jabbar")) # head <-> milad <-> Arya
print("Current size of list is:", list.size())
```
