# Problem 1: Maximum Contiguous Subarray

  Given a sequence of real numbers, we want to find a contiguous subarray with the maximum sum among all possible contiguous subarrays.  
  
  
  *Note:* It is strongly recommended that you spend a few minutes attempting to solve this problem on your own before reading the solution.  
  
  # First Solution

 The first approach is to consider all possible contiguous subarrays, sum their elements, and return the subarray with the maximum sum.

```python
import sys

n = 10
arr = [1, 2, 31, -3, 51, 12, -102, 1, 31, 1]
max_sum = -sys.maxsize

for i in range(n):
    for j in range(i, n):
        total_sum = 0
        for k in range(i, j + 1):
            total_sum += arr[k]
        if total_sum > max_sum:
            max_sum = total_sum

print(max_sum)
```

### Time Complexity Analysis of the First Solution  

Analyzing the time complexity shows this takes $O(n^3)$ time. Choosing the start and end indices takes $\frac{n \times (n+1)}{2}$ operations, and iterating through elements between these indices to sum them takes up to $O(n)$ operations. Thus, the overall complexity is $O(n^3)$. Is there a faster way?  

# Second Solution
  We can optimize this by avoiding recomputing sums from scratch. When calculating the sum from index $i$ to $j$, instead of re-adding all elements, we can simply add element $j$ to the previously calculated sum from $i$ to $j-1$.

```python
import sys

n = 10
arr = [1, 2, 31, -3, 51, 12, -102, 1, 31, 1]
max_sum = -sys.maxsize

for i in range(n):
    total_sum = 0
    for j in range(i, n):
        total_sum += arr[j]
        if total_sum > max_sum:
            max_sum = total_sum

print(max_sum)
```

### Time Complexity Analysis of the Second Solution  
This approach runs in $O(n^2)$ time. Choosing start and end indices requires $\frac{n \times (n+1)}{2}$ operations, but summing elements now takes $O(1)$ time per step. Can we do better?

# Third Solution
  Iterate through the array from the beginning. Maintain a running sum. For each element, add its value to the running sum. If the sum is positive, compare it with the global maximum sum found so far and update it if higher. If the running sum becomes negative, reset it to 0 and move to the next element. Repeat until the end of the array.

```python
import sys

n = 10
arr = [1, 2, 31, -3, 51, 12, -102, 1, 31, 1]
max_sum = -sys.maxsize

total_sum = 0
for i in range(n):
    if total_sum < 0:
        total_sum = 0
    total_sum += arr[i]
    if total_sum > max_sum:
        max_sum = total_sum

print(max_sum)
```

### Time Complexity Analysis of the Third Solution

  This algorithm runs in $O(n)$ time because it iterates over the array once, performing constant $O(1)$ operations (addition, comparison, assignment) per element. We reduced an initial $O(n^3)$ problem to a linear $O(n)$ time complexity.  

# Problem 2: 3-Sum
  Given a set of $n$ real numbers, determine whether there exist three elements in the set whose sum is zero.  
  
  *Note:* It is strongly recommended that you try solving this on your own before continuing.  
  
  # First Solution

   The brute-force approach checks every triplet of elements, calculates their sum, and checks if it equals zero.

```python
def solve_3sum():
    n = 10
    arr = [1, 2, 31, -3, 51, 12, -102, 1, 31, 1]

    for i in range(n):
        for j in range(n):
            for k in range(n):
                if i != j and j != k and k != i and arr[i] + arr[j] + arr[k] == 0:
                    print("Yes")
                    return
    print("No")

solve_3sum()
```

### Time Complexity Analysis of the First Solution

  Iterating through all elements in three nested loops results in $n^3$ operations, giving an $O(n^3)$ time complexity.

# Second Solution

  We can eliminate duplicate checks by restricting loop indices so $i < j < k$.

```python
def solve_3sum():
    n = 10
    arr = [1, 2, 31, -3, 51, 12, -102, 1, 31, 1]

    for i in range(n):
        for j in range(i + 1, n):
            for k in range(j + 1, n):
                if arr[i] + arr[j] + arr[k] == 0:
                    print("Yes")
                    return
    print("No")

solve_3sum()
```

### Time Complexity Analysis of the Second Solution

  The number of operations corresponds to choosing 3 items out of $n$, which is $\frac{n \times (n-1) \times (n-2)}{6}$ operations. The overall time complexity remains $O(n^3)$.

# Third Solution

  First, sort the array. Fix the first element using an index $i$ from $0$ to $n-2$. Use two pointers for the remaining two elements: set $j = i + 1$ (left pointer) and $k = n - 1$ (right pointer). Check the sum of arr[i] + arr[j] + arr[k]:
  

   * If sum == 0, a valid triplet is found.
   * If sum > 0, decrement $k$.  
   * If sum < 0, increment $j$.  




```python
def solve_3sum():
    n = 10
    arr = [1, 2, 31, -3, 51, 12, -102, 1, 31, 1]

    arr.sort()
    for i in range(n - 1):
        j = i + 1
        k = n - 1
        while j < k:
            total_sum = arr[i] + arr[j] + arr[k]
            if total_sum == 0:
                print("Yes")
                return
            elif total_sum > 0:
                k -= 1
            else:
                j += 1
    print("No")

solve_3sum()
```

### Time Complexity Analysis of the Third Solution
  Sorting takes $O(n \log n)$ time. The outer loop runs $n-1$ times, and the two-pointer inner loop processes at most $n-1$ elements total per iteration. The total steps for the loops are bounded by $(n-1)^2$, yielding an overall time complexity of $O(n^2)$.
